```python
# Importar libreria Geopandas
import geopandas as gpd
import urllib.request
```

```python
url = "https://www.ideandalucia.es/services/DERA_g12_servicios/wfs?service=wfs&request=getcapabilities"

gdf1 = gpd.read_file(
    url, 
    layer='DERA_g12_servicios:g12_01_CentroSalud', 
    engine="pyogrio", 
    encoding="utf-8"
)
```

```python
gdf1
```

```python
# Borrrar columnas no necesarias
gdf1 = gdf1.drop(columns=[ 'gml_id', 'id_dera', 'nombre', 'direccion', 'telefono', 'web', 'localidad', 'cod_mun',])
```
